**P16. Mini Data Validator**
You receive one record as a dictionary:
```python
record = {
    "id":         "101",
    "name":       "Alice",
    "salary":     "95000",
    "department": "Engineering",
    "active":     "True"
}
```
Write code that:
1. Checks all required keys are present (`"id"`, `"name"`, `"salary"`, `"department"`) using `in`
2. Casts `"id"` to `int`, `"salary"` to `float`, `"active"` to `bool`
3. Checks if `salary > 80000` and `active is True`
4. Prints a summary: `"Record 101 — Alice | Salary: 95000.0 | Active: True | High earner: True"`

> This pattern — validate schema → cast types → apply business rules — is the core of every data ingestion pipeline.


````python
record = {
    "id":         "101",
    "name":       "Alice",
    "salary":     "95000",
    "department": "Engineering",
    "active":     "True"
}

# 1. Check required keys
required_keys = ["id", "name", "salary", "department"]

if all(key in record for key in required_keys):
    print("All required keys are present")
else:
    print("Missing required keys")


# 2. Casts `"id"` to `int`, `"salary"` to `float`, `"active"` to `bool`

record["id"] = int(record["id"])
record["salary"] = float(record["salary"])
record["active"] = record["active"] == "True"

print(record["id"])
print(record["salary"])
print(record["active"])

# 3. Checks if `salary > 80000` and `active is True`

high_earner = record["salary"] > 80000 and record["active"] is True

print(high_earner)


# 4. Prints a summary: `"Record 101 — Alice | Salary: 95000.0 | Active: True | High earner: True"`


print(
    f"Record {record['id']} — {record['name']} | "
    f"Salary: {record['salary']} | "
    f"Active: {record['active']} | "
    f"High earner: {high_earner}"
)
````
